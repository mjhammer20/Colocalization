# Helper Notebook To Create Ancestry Keep File
This notebook is intended to help create an ancestry keep file for the reference panel that will be used to compute LD in the colocalization worklfow. The user will need to specify the following:
- Path to ref panel directory
- Name of ancestry file for ref panel
- Target ancestry
- Name of ancestry keep file output

## Imports

```python
import pandas as pd
from pathlib import Path
```

## Configuration

```python
ref_panel_dir = "/mnt/working/ref_panels/1kg/ref_panel"
ancestry_file = "ref_panel_ancestry_updated.txt"
target_ancestry = "EUR"
ancestry_keep_file = "ref_panel_ancestry_keep.txt"
```

## Load Input

```python
ancestry = pd.read_csv(Path(ref_panel_dir, ancestry_file), sep="\t", header=None)
```

## Filter to Target Ancestry

```python
target_ancestry_samples = ancestry[ancestry.iloc[:,2] == target_ancestry]
```

## Write Ancestry Keep File

```python
target_ancestry_samples.to_csv(Path(ref_panel_dir, ancestry_keep_file), sep="\t", index=False, header=False)
```
