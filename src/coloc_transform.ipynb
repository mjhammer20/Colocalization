```python
# Imports
import pandas as pd
from pathlib import Path
import coloc_transform
```

```python
# Load Summary Statistics Data
input_file_path = Path("/mnt/working/locus_reports/qtl/eqtl/eqtlgen/cis/2019-12-11-cis-eQTLsFDR0.05-ProbeLevel-CohortInfoRemoved-BonferroniAdded.txt.gz")  # Replace with your actual file path
header_lines = 0  # Adjust based on your file's header lines
df = coloc_transform.load_input_data(input_file_path, header_lines)

# Print the shape of the DataFrame
print(f"DataFrame shape: {df.shape}")

# Print the column names of the DataFrame
print(f"Column names: {df.columns.tolist()}")

# Display the first few rows of the DataFrame
df.head()
```

```python
# Configuration
chr_col = "chromosome"
pos_col = "snpLocation"
rsid_col = "snpid" # Set to None if not available
gene_id_col = "gene" # Set to None if not available
effect_allele_col = "A2" # Set to None if not available
non_effect_allele_col = "A1" # Set to None if not available
n_col = None # Set to None if not available
maf_col = "A2freq" # Set to None if not available
mac_col = None # Set to None if not available
p_col = "pvalue" # Set to None if not available
statistic_col = "statistic" # Set to None if not available
se_col = None # Set to None if not available
beta_col = "beta" # Set to None if not available
var_beta_col = None # Set to None if not available
sdY_col = None # Set to None if not available
```

```python
df[chr_col] = df[chr_col].astype(str).apply(coloc_transform.normalize_chromosome)
```

```python
expanded_ranges_fp = Path('/mnt/working/locus_reports/expanded_locus_ranges_by_chromosome.json')
expanded_ranges = coloc_transform.load_expanded_ranges(expanded_ranges_fp)
# Filter the DataFrame by expanded ranges
filtered_df = coloc_transform.filter_by_expanded_ranges(
    df=df,
    chr_col=chr_col,
    pos_col=pos_col,
    expanded_ranges=expanded_ranges
)

# Print the shape of the DataFrame
print(f"DataFrame shape: {filtered_df.shape}")

# Print the column names of the DataFrame
print(f"Column names: {filtered_df.columns.tolist()}")

# Display the first few rows of the DataFrame
filtered_df.head()
```

```python
annotated_df = coloc_transform.annotate_sum_stats(
    df=filtered_df,
    chr_col=chr_col,
    pos_col=pos_col,
    rsid_col=rsid_col,
    effect_allele_col=effect_allele_col,
    non_effect_allele_col=non_effect_allele_col,
    n_col=n_col,
    maf_col=maf_col,
    mac_col=mac_col,
    p_col=p_col,
    statistic_col=statistic_col,
    se_col=se_col,
    beta_col=beta_col,
    var_beta_col=var_beta_col,
    sdY_col=sdY_col
)
```

```python
transformed_df = coloc_transform.transform_sum_stats(
        df=annotated_df,
        chr_col=chr_col,
        pos_col=pos_col,
        variant_id_col='Variant_ID',
        rsid_col=rsid_col or 'rsID',
        gene_id_col=gene_id_col,
        effect_allele_col=effect_allele_col or 'dbSNP_alt',
        non_effect_allele_col=non_effect_allele_col or 'dbSNP_ref',
        p_col=p_col or 'P',
        beta_col=beta_col or 'BETA',
        se_col=se_col or 'SE',
        n_col=n_col or 'N',
        maf_col=maf_col or 'MAF',
        var_beta_col=var_beta_col or 'VARBETA',
        sdY_col=sdY_col or 'SDY'
    )

transformed_df.head()
```

```python
# Save the transformed data to a new file
output_dir = "/mnt/working/locus_reports/qtl/eqtl/coloc_ready" # Replace with your desired output file path
output_fn = input_file_path.stem + ".coloc_ready.tsv"
output_fp = Path(output_dir) / output_fn
transformed_df.to_csv(output_fp, sep="\t", index=False)
```
